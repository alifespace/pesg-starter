## 字典基础
1. 字典是一种键值对容器，使用 {} 表示；
2. key值不能重复，重复的话，后面会覆盖前面；
3. ke值一般是字符串，数值，元组；
4. value通常可以是字符串、数字、列表、字典；
4. 现代python中，键值对通常保持插入顺序；

如果存的是一组同类数据，通常用列表。如果你存的是一个对象的多个属性，通常用字典。

## 常见操作
### 定义空字典
1. data = {}
2. data = dict()

### 取值
1. student['name']
2. student.get('name')
3. name in student
4. len(student)

### 新增、修改、删除
1. student['score'] = 95
2. del student['score']
3. student.pop['city']
4. student.clear()

### 常见遍历方式
1. for key in student:
2. for value in student.values():
3. for key, value in student.items():
