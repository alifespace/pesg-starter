```python
scores =["张三, 78", "李四, 92", "王五, 85", "赵六, 60"]
total = 0
max_score = 0
top_student = ""

for score in scores:
    name, score_value = score.split(", ")
    total += int(score_value)
    if int(score_value) >= max_score:
        max_score = int(score_value)
        top_student = name.strip()

avg_score = total / len(scores)
print("平均分:", avg_score)
print("最高分:", max_score)
print("最高分学生:", top_student)
```
