```python
# 在不改原函数代码的前提下，给这个函数额外加一点能力。把“核心逻辑”和“附加逻辑”分开
# 装饰器本质上是一个函数，它接收另一个函数作为参数，并返回一个增强后的新函数

def say_hello():
    print('你好')

def decorator(func):
    def wrapper():
        print('开始执行函数')
        func()
        print('结束执行函数')
    return wrapper

print(id(say_hello), say_hello.__name__)  # say_hello函数的内存地址

# 把原来的 say_hello 函数交给 decorator
# decorator 返回一个增强后的新函数 再让变量名 say_hello 指向这个增强后的版本
say_hello = decorator(say_hello)  # say_hello函数被装饰器装饰了，say_hello = wrapper
print(id(say_hello), say_hello.__name__)  # say_hello函数的内存地址
# 查看变量的内存地址
say_hello()
```

```python
@decorator
def say_hello2():
    print("你好")

say_hello2()
```

```python
def decorator(func):
    def wrapper(a, b):
        print('开始执行')
        func(a, b)
    return wrapper

@decorator
def add(a, b):
    print(a + b)

add(3, 5)
```

```python
def log_decorator(func):
    def wrapper():
        print('函数开始执行')
        func()
        print('函数执行结束')
    return wrapper

@log_decorator
def test():
    print('这里是函数主体')

test()
```

```python
print("【1】程序开始")

def common_decorator(func):
    print("【2】装饰器函数被调用 （定义时执行）")
    print(f"【3】接收到原函数: {func.__name__}")
    
    def wrapper(*args, **kwargs):
        print("【6】wrapper 被执行（调用时执行）")
        res = func(*args, **kwargs)
        return res
    
    print("【4】装饰器返回 wrapper")
    return wrapper

print("【5】开始定义被装饰的函数")

@common_decorator
def add(a, b):
    print("【7】原 add 函数执行")
    return a + b

print("【8】函数定义完成")
print("【9】准备调用 add(10, 20)")

result = add(10, 20)

print("【10】调用完成")
print(f"【11】结果: {result}")
```
