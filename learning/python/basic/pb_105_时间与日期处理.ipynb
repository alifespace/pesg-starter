```python
import time
from datetime import date, datetime, timezone, timedelta

now = datetime.now(timezone.utc)

print(now, type(now))

today = date.today()
print(today)

dt = datetime(2026, 3, 29, 23, 30, 0)
print(dt)

d = date(2023, 4, 20)
print(d)

print(now.strftime("%Y-%m-%d"))
print(now.strftime("%H:%M:%S"))

text = "2026-03-29 15:20:30"
dt = datetime.strptime(text, "%Y-%m-%d %H:%M:%S")

print(dt)
```

```python
start = datetime.now()
time.sleep(1)
end = datetime.now()

diff = end -start

print(diff, type(diff))
print(diff.total_seconds())
print(diff.days)
```

```python
now = datetime.now()

print(now, now + timedelta(hours=2))
```

```python
expire_text = "2026-04-05 23:59:59"
expire_time = datetime.strptime(expire_text, "%Y-%m-%d %H:%M:%S")

now = datetime.strptime("2026-03-29 12:00:00", "%Y-%m-%d %H:%M:%S")

if now <= expire_time:
    diff = expire_time - now
    print("会员仍然有效")
    print(f"距离到期还有 {diff.days} 天")
else:
    print("会员已过期")
```
