```python
import re

text = "用户A电话是13812345678, 用户B电话是15600001111"
print(re.findall(r"\d{11}", text))
print(re.search(r"\d{11}", text).group())
print(re.findall(r"1[3-9]\d{9}", text))
```

```python
text = "今天是2026-03-29"
result = re.search(r"(\d{4})-(\d{2})-(\d{2})", text)
print(result.group(0))
print(result.group(1))
print(result.group(2))
print(result.group(3))
```

```python
text = "邮箱是abc@qq.com"
print(re.match(r"[\w.-]+@[\w.-]+\.\w+", text).group())
print(re.search(r"[\w.-]+@[\w.-]+\.\w+", text).group())
```

```python
phone_pattern = re.compile(r"1[3-9]\d{9}")
email_pattern = re.compile(r"[\w.-]+@[\w.-]+\.\w+")
date_pattern = re.compile(r"\d{4}[-/]\d{2}[-/]\d{2}")
```
