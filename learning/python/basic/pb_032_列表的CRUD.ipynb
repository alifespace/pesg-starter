## 列表的基础操作CRUD
1. 新增：append(), insert(), extend();
2. 删除：remove(), pop(), del, clear();
3. 修改：lst[下标] = 新值；
4. 查询：in, count(), index(), len()

remove(): 按值删除；
pop(): 按位置删除；
del 列表[下标]: 也是按照位置删除 

```python
cart = ['鼠标', '键盘', '耳机']
cart.append('显示器')
cart.remove('键盘')
cart[0] = '无线鼠标'
print(cart, '耳机' in cart)
print(len(cart))
```

```python
cart.insert(1, '苹果')
print(cart)
new_cart = ['箱子', '耳环']
cart.extend(new_cart)
print(cart)
```

```python
cart.remove('箱子')
print(cart)
cart.pop(1)
print(cart)
```

```python
print('耳机' in cart)
print(cart.count('耳机'))
```

```python
print(len(cart))
print(cart.index('耳机'))
```
