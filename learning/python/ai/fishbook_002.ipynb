```python
# 感知机（perceptron）的实现
import numpy as np

# 与门实现
# def AND(x1, x2):
#     w1, w2, theta = 0.5, 0.5, 0.7
#     tmp = x1 * w1 + x2 * w2
#     if tmp <= theta:
#         return 0
#     elif tmp > theta:
#         return 1

def AND(x1, x2):
    x = np.array([x1, x2])
    w = np.array([0.5, 0.5])
    b = -0.7
    tmp = np.sum(w * x) + b
    if tmp <= 0:
        return 0
    else:
        return 1

print(AND(0, 0))  # 输出 0
print(AND(1, 0))  # 输出 0
print(AND(0, 1))  # 输出 0
print(AND(1, 1))  # 输出 1
```

```python
# 与非门实现
def NAND(x1, x2):
    x1 = np.array([x1, x2])
    w = np.array([-0.5, -0.5])
    b = 0.7
    tmp = np.sum(w * x1) + b
    if tmp <= 0:
        return 0
    else:
        return 1

def OR(x1, x2):
    x1 = np.array([x1, x2])
    w = np.array([0.5, 0.5])
    b = -0.2
    tmp = np.sum(w * x1) + b
    if tmp <= 0:
        return 0
    else:
        return 1

print(NAND(0, 0))  # 输出 1
print(NAND(1, 0))  # 输出 1
print(NAND(0, 1))  # 输出 1
print(NAND(1, 1))  # 输出 0

print(OR(0, 0))  # 输出 0
print(OR(1, 0))  # 输出 1
print(OR(0, 1))  # 输出 1
print(OR(1, 1))  # 输出 1
```

```python
def XOR(x1, x2):
    s1 = NAND(x1, x2)
    s2 = OR(x1, x2)
    y = AND(s1, s2)
    return y

print(XOR(0, 0))
print(XOR(1, 1))
print(XOR(1, 0))
print(XOR(0, 1))
```
