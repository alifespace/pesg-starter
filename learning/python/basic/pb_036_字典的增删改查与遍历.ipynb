```python
student = {'name':'张三', 'age':18, 'gender':'男', 'score':95}

# 查
print(student['name'], student.get('age', '年龄不存在'), student.get('address', '地址不存在'))

# 增
student['address'] = '北京市'
print(student)

# 删
del student['address']
print(student)
student['address'] = '北京市'
print(student.pop('address', '地址不存在'))
print(student.pop('address', '地址不存在'))

print(len(student), student)

# 更新
student.update({'age':20, 'score':96, 'city':'上海'})
print(student)
```
