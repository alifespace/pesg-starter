## 集合基本概念

集合是一种装多个元素的容器。特点：

1. 元素不重复；
2. 无序；
3. 适合判断成员、去重、做交并集；
4. 集合的元素必须可哈希，所以是数字、字符串、元组，列表不能作为集合的元素；
4. 用{}表示集合；
5. 空集合不能写成 {}，因为表示的是空字典；

```python
set1 = {0, 1, 2, 3, 4, (1, 2)}
set2 = {3, 4, 5, 6} 
set3 = set() # This will raise a TypeError because dictionaries are unhashable and cannot be added to a set
print(set1, type(set1))
```

```python
set3.add(1)
print(set3, type(set3))
set3.update([2, 3, 4])
print(set3, type(set3))
set3.remove(2)
print(set3, type(set3))
set3.discard(3)
print(set3, type(set3))
```

```python
print(set1, set2)
print(set1 & set2)
print(set1.intersection(set2))

print(set1 | set2)
print(set1.union(set2))

# 差集
print(set1 - set2)
print(set2 - set1)
print(set1.difference(set2))

# 对称差集
print(set1 ^ set2)
print(set1.symmetric_difference(set2))
```
