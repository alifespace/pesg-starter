```python
def add(*args):
    print(args)
    total = 0
    for num in args:
        total += num
    return total

total = add(1, 2, 3, 4)
print(total)
```

```python
# **kwargs: 关键字参数数量不固定
# 它会把传进来的多个关键字参数，自动打包成一个字典
def show_info(**kwargs):
    for key, value in kwargs.items():
        print(key, value)

show_info(name='张三', age=18, city='北京')
```

```python
def demo(*args, **kwargs):
    print(args)
    print(kwargs)

demo(1, 2, 3, name='张三', age=18)
```
