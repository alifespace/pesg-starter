```python
# map: 就是把同一个处理规则，批量作用到一组数据上
# map(函数, 可迭代对象)

nums = [1, 2, 3, 4, 5]

result = map(lambda x: x * 2, nums)
print(result, type(result), list(result))  # 输出: <map object at 0x...> <class 'map'>
```

```python
words = ['python', 'java', 'go']

result = map(lambda x: x.upper(), words)
print(result, type(result), list(result))  # 输出: <map object at 0
```

```python
nums = [10, 20, 30]

result = map(str, nums)
print(list(result))
```

```python
nums = [1, 2, 3, 4, 5, 6]

print([x for x in nums if x % 2 == 0])  # 输出: [2, 4, 6]
result = filter(lambda x: x % 2 == 0, nums)
print(result, type(result), list(result))  # 输出: <filter object at 0
```

```python
from functools import reduce

nums = [1, 2, 3, 4, 5]

# reduce(函数, 可迭代对象)
result = reduce(lambda x, y: x + y, nums)
print(result)  # 输出: 15
```
