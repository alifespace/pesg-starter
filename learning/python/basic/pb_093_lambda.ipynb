```python
# lambda 参数: 返回值表达式
# lambda 把它作为一个临时函数，传给别的函数使用
# lambda 的真正价值不在“短”，而在“临时传入”

double = lambda x: x * 2

print(double(5))  # 输出: 10
print(double(10)) # 输出: 20
print(type(double))  # 输出: <class 'function'>
```

```python
students = [('张三',95),
            ('李四',88),
            ('王五',91)
]

students.sort(key = lambda x: x[1])
print(students)
```

```python
words = ['python', 'go', 'banana', 'cat']

words.sort(key = lambda x: len(x))
print(words)
```

```python
check = lambda x: '及格' if x >= 60 else '不及格'
print(check(75))  # 输出: 及格
print(check(50))  # 输出: 不及格
```
