```python
import matplotlib.pyplot as plt
import numpy as np

# 设置中文字体（macOS 推荐 PingFang SC，Linux 可用 'WenQuanYi Micro Hei'）
plt.rcParams['font.sans-serif'] = ['PingFang SC', 'WenQuanYi Micro Hei', 'SimHei', 'Microsoft YaHei']
plt.rcParams['axes.unicode_minus'] = False  # 解决负号显示为方块的问题
```

```python
# 定义一个类

class Man:
    def __init__(self, name):
        self.name = name
        print("Initializing Man")

    def hello(self):
        print("Hello" + self.name + "!")

    def goodbye(self):
        print("Goodbye" + self.name + "!")

m = Man("Droid")
m.hello()
m.goodbye()
```

```python
x = np.array([1.0, 2.0, 3.0])
y = np.array([4.0, 5.0, 6.0])
print(x)
print(type(x))

print(x / y)
print(x * 2)
```

```python
A = np.array([[1, 2], [3, 4]])
B = np.array([[3, 0], [0, 6]])
print(A)

print(A.shape)
print(A.dtype)

print (A + B)

print(A * B)  # 对应元素相乘, 不是矩阵的乘法
print(A * 10)
print(A * np.array([10, 20]))  # 广播机制
```

```python
x = np.array([[1, 2], [3, 4], [5, 6]])
for row in x:
    print(row)

X = x.flatten()  # 将多维数组展平为一维数组
print(X)
print(X[np.array([0, 2, 4])])  # 使用整数数组进行索引
print(X > 3)  # 返回布尔数组
print(X[X > 3])  # 使用布尔数组进行索引
```

```python
x = np.arange(0, 6, 0.1)
y = np.sin(x)

plt.plot(x, y)
plt.title('正弦函数图像')
plt.show()
```

```python
x = np.arange(0, 6, 0.1)
y1 = np.sin(x)
y2 = np.cos(x)

# 绘制图形
plt.plot(x, y1, label='sin(x)', color= 'blue', linestyle='-')
plt.plot(x, y2, label='cos(x)', color= 'red', linestyle='--')
plt.xlabel('x')
plt.ylabel('y')
plt.title('正弦函数与余弦函数图像')

plt.legend()  # 显示图例


plt.show()
```
