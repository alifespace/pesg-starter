## JSON 基础

JSON 很像字典。 确实，它们长得非常像，但不是一回事。JSON 是一种数据格式。 字典是 Python 里的数据类型。

1. JSON 里的键，必须用双引号。 JSON 里的布尔值是 true 和 false。 JSON 里的空值是 null；
2. python 里对应的是：True，False，None;



```python
import json

json_data = '''
{
    "username": "tom",
    "age": 20,
    "city": "上海"
}
'''

# 只要记住后面多了个 s，可以把它理解成 string 的意思，基本就不容易混。
data = json.loads(json_data)
print(data, type(data))
```

```python
with open('user.json', 'r', encoding='utf-8') as f:
    data = json.load(f)

print(data, type(data))
```

```python
user = {
    "name": "小李",
    "age": 22,
    "city": "北京"
}

jason_data = json.dumps(user, ensure_ascii=False, indent=4)
print(jason_data, '\n',type(jason_data))
```

```python
users = [
    {'name': '张三', 'age': 18},
    {'name': '李四', 'age': 20},
    {'name': '王五', 'age': 22}
]

with open('users.json', 'w', encoding='utf-8') as f:
    json.dump(users, f, ensure_ascii=False, indent=4)
```

```python
response = '''
{
    "code": 200,
    "data": [
        {"name": "张三", "scores": [80, 85, 90]},
        {"name": "李四", "scores": [78, 88, 82]},
        {"name": "王五", "scores": [92, 95, 91]}
    ]
}
'''

data = json.loads(response)
print(data, type(data))
print(data['data'])
for student in data['data']:
    name = student['name']
    scores = student['scores']
    total = 0
    for score_value in scores:
        total += int(score_value)
    avg = total / len(scores)
    print(f"{name}的平均分是: {avg}")
```
