```python
#{键表达式: 值表达式 for 变量 in 可迭代对象}

words = ['apple', 'banana', 'cat']

print({word: len(word) for word in words})
```

```python
students = ['张三', '李四', '王五']
scores = [95, 88, 91]

print({students[i]: scores[i] for i in range(len(students)) })
```

```python
# {键表达式: 值表达式 for 变量 in 可迭代对象 if 条件}

words = ['cat', 'python', 'book', 'banana']

print({word: len(word) for word in words if len(word) > 3})
```

```python
info = {'a': 1, 'b': 2, 'c': 3}

print({v: k for k, v in info.items()})
```
