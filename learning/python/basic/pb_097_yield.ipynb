```python
# return 是函数彻底结束，并把结果交出去。yield 是先交一个值出去，但函数并不真正结束，而是暂停在这里
# 生成器函数的核心不是“调用就执行完”，而是“创建一个可逐步执行的过程”

def simple_gen():
    yield 1
    yield 2
    yield 3

g = simple_gen()

print(next(g))
print(next(g))
print(next(g))
```

```python
def square_gen(n):
    for i in range(1, n):
        yield i * i

g = square_gen(5)
print(next(g))
print(next(g))
print(next(g))
print(next(g))
```

```python
def demo():
    print('start')
    yield 1
    print('continue')
    yield 2
    print('end')

g = demo()
print(next(g))
print(next(g))
print(next(g))  # StopIteration
```

```python
def word_gen(text):
    for word in text.split():
        yield word

for w in word_gen('Hello World!'):
    print(w)
```
