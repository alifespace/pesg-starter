```python
nums = [1, 2, 3, 4, 5]

result = []
for num in nums:
    result.append(num ** 2)

print(result)  # 输出: [1, 4, 9, 16, 25]
```

```python
# 要从一组旧数据，快速生成一组新列表
# [表达式 for 变量 in 可迭代对象]
result = [num ** 2 for num in nums]
print(result)  # 输出: [1, 4, 9, 16, 25]

print([x * 2 for x in [1, 2, 3]])
```

```python
words = ['python', 'java', 'go']

print([word.upper() for word in words]) # 输出: ['PYTHON', 'JAVA', 'GO']
```

```python
# [表达式 for 变量 in 可迭代对象 if 条件]
nums = [1, 2, 3, 4, 5, 6]

print([num for num in nums if num % 2 ==0])
```

```python
words =['apple', 'dog', 'banana', 'cat']

print([word for word in words if 'a' in word])  # 输出: ['apple', 'banana', 'cat']
```
