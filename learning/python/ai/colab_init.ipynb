```python
from flask import Flask

app = Flask(__name__)

@app.route("/")
def home():
    return """
    <html>
        <head>
            <title>My Colab Website2 :)</title>
        </head>
        <body>
            <h1>Hello from Google Colab2, Today!</h1>
            <p>This page is running with Flask.</p>
        </body>
    </html>
    """

@app.route("/about")
def about():
    return """
    <h1>About</h1>
    <p>This is my second Flask page for colab2.</p>
    """

import threading
from werkzeug.serving import make_server

class FlaskServer(threading.Thread):
    def __init__(self, app, host="0.0.0.0", port=5002):
        super().__init__()
        self.server = make_server(host, port, app)
        self.ctx = app.app_context()
        self.ctx.push()

    def run(self):
        self.server.serve_forever()

    def shutdown(self):
        self.server.shutdown()
        self.ctx.pop()

flask_server1 = FlaskServer(app)
flask_server1.start()

print("Flask started")
```

```python
flask_server1.shutdown()
flask_server1.join()

print("Flask stopped")
```
